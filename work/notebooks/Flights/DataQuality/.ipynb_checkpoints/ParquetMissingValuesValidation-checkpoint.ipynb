# 🔍 Validation des Valeurs Manquantes dans les Parquets

Ce notebook permet de valider que les fichiers parquet générés contiennent correctement :
- ✅ Les flags de missing par heure (`_missing_h1` à `_missing_h7`)
- ✅ Les compteurs de missing pour features agrégées (`_missing_count`)
- ✅ Les valeurs sentinelles (-999, "MISSING")
- ✅ Aucune valeur NULL restante

---

## 📦 1. Configuration et Imports

```scala
import $ivy.`org.apache.spark::spark-sql:3.5.0`
import org.apache.spark.sql.{SparkSession, DataFrame}
import org.apache.spark.sql.functions._

// Créer la session Spark
val spark = SparkSession.builder()
  .appName("ParquetMissingValuesValidation")
  .master("local[*]")
  .config("spark.driver.memory", "4g")
  .config("spark.sql.shuffle.partitions", "8")
  .getOrCreate()

println("✅ Spark session créée")
```

## 📂 2. Chargement des Parquets

```scala
// Chemins vers les parquets
val experimentName = "Experience-local-D-60-7-7"
val basePath = "/output"

val trainPath = s"$basePath/$experimentName/data/join_exploded_train_prepared.parquet"
val testPath = s"$basePath/$experimentName/data/join_exploded_test_prepared.parquet"

println(s"📁 Train path: $trainPath")
println(s"📁 Test path:  $testPath")
```

```scala
// Charger les datasets
val trainDF = spark.read.parquet(trainPath)
val testDF = spark.read.parquet(testPath)

val trainCount = trainDF.count()
val testCount = testDF.count()

println(s"✅ Train dataset: ${trainCount} lignes, ${trainDF.columns.length} colonnes")
println(s"✅ Test dataset:  ${testCount} lignes, ${testDF.columns.length} colonnes")
```

## 🔍 3. Validation des Flags par Heure

```scala
// Identifier les colonnes de flags
val hourlyFlagCols = trainDF.columns.filter(_.matches(".+_missing_h\\d+"))

println(s"\n📊 Flags par heure trouvés: ${hourlyFlagCols.length}")
println("\nExemples:")
hourlyFlagCols.sorted.take(10).foreach(c => println(s"  - $c"))

if (hourlyFlagCols.length > 10) {
  println(s"  ... et ${hourlyFlagCols.length - 10} autres")
}
```

```scala
// Statistiques sur les données manquantes (TRAIN)
println("\n📊 Statistiques des données manquantes par heure (TRAIN):")
println("=" * 80)

hourlyFlagCols.sorted.take(10).foreach { colName =>
  val missingCount = trainDF.filter(col(colName) === 1).count()
  val missingPct = (missingCount.toDouble / trainCount * 100)
  println(f"$colName%-60s: $missingCount%,8d ($missingPct%5.2f%%)")
}

if (hourlyFlagCols.length > 10) {
  println(s"... et ${hourlyFlagCols.length - 10} autres")
}
```

## 🧮 4. Validation des Features Agrégées

```scala
// Identifier les features agrégées
val aggMethods = Seq("Sum", "Avg", "Max", "Min", "Std")
val aggFeatureCols = trainDF.columns.filter { colName =>
  aggMethods.exists(method => colName.contains(s"_$method"))
}.filterNot(_.contains("_missing_count"))

println(s"\n📊 Features agrégées trouvées: ${aggFeatureCols.length}")
println("\nExemples:")
aggFeatureCols.sorted.take(10).foreach(c => println(s"  - $c"))
```

```scala
// Vérifier la présence des compteurs
println("\n🔍 Vérification des compteurs _missing_count:")
println("=" * 80)

aggFeatureCols.sorted.foreach { aggCol =>
  val expectedCounterCol = s"${aggCol}_missing_count"
  val hasCounter = trainDF.columns.contains(expectedCounterCol)
  val status = if (hasCounter) "✓" else "✗ MANQUANT"
  println(f"$aggCol%-60s: $status")
}
```

```scala
// Statistiques sur les compteurs (TRAIN)
val counterCols = trainDF.columns.filter(_.endsWith("_missing_count"))

println("\n📊 Statistiques des compteurs de missing (TRAIN):")
println("=" * 80)

counterCols.sorted.foreach { counterCol =>
  val stats = trainDF.agg(
    avg(col(counterCol)).as("avg_count"),
    max(col(counterCol)).as("max_count")
  ).first()

  val avgCount = stats.getAs[Double]("avg_count")
  val maxCount = stats.getAs[Long]("max_count")

  println(f"$counterCol%-60s: avg=$avgCount%5.2f, max=$maxCount%2d")
}
```

## 🏷️ 5. Validation des Valeurs Sentinelles

```scala
// Colonnes numériques (double/int)
val numericCols = trainDF.schema.fields
  .filter(f => f.dataType.typeName == "double" || f.dataType.typeName == "integer" || f.dataType.typeName == "long")
  .map(_.name)
  .filterNot(_.contains("_missing")) // Exclure flags et compteurs

println(s"\n📊 Colonnes numériques: ${numericCols.length}")
```

```scala
// Vérifier les sentinelles -999 (TRAIN)
println("\n🔍 Colonnes avec sentinelles numériques -999 ou -999.0 (TRAIN):")
println("=" * 80)

val colsWithSentinels = numericCols.flatMap { colName =>
  val sentinelCount = trainDF.filter(
    col(colName) === -999.0 || col(colName) === -999
  ).count()

  if (sentinelCount > 0) {
    val sentinelPct = (sentinelCount.toDouble / trainCount * 100)
    Some((colName, sentinelCount, sentinelPct))
  } else {
    None
  }
}.sortBy(-_._2)

colsWithSentinels.take(15).foreach { case (colName, count, pct) =>
  println(f"$colName%-60s: $count%,8d ($pct%5.2f%%)")
}

if (colsWithSentinels.length > 15) {
  println(s"... et ${colsWithSentinels.length - 15} autres colonnes avec sentinelles")
}

if (colsWithSentinels.isEmpty) {
  println("ℹ️  Aucune sentinelle -999 détectée (normal si pas de données manquantes)")
}
```

```scala
// Vérifier les sentinelles "MISSING" (TRAIN)
val stringCols = trainDF.schema.fields
  .filter(_.dataType.typeName == "string")
  .map(_.name)

println("\n🔍 Colonnes avec sentinelles string \"MISSING\" (TRAIN):")
println("=" * 80)

val colsWithMissing = stringCols.flatMap { colName =>
  val missingCount = trainDF.filter(col(colName) === "MISSING").count()

  if (missingCount > 0) {
    val missingPct = (missingCount.toDouble / trainCount * 100)
    Some((colName, missingCount, missingPct))
  } else {
    None
  }
}.sortBy(-_._2)

colsWithMissing.take(10).foreach { case (colName, count, pct) =>
  println(f"$colName%-60s: $count%,8d ($pct%5.2f%%)")
}

if (colsWithMissing.isEmpty) {
  println("ℹ️  Aucune sentinelle MISSING détectée (normal si pas de données manquantes)")
}
```

## 📋 6. Échantillon de Données

```scala
// Sélectionner des colonnes intéressantes
val sampleCols = Seq(
  trainDF.columns.find(_.contains("HourlyPrecip_Sum")),
  trainDF.columns.find(_.contains("HourlyPrecip_Sum_missing_count")),
  trainDF.columns.find(_.contains("press_change_abs_Max")),
  trainDF.columns.find(_.contains("press_change_abs_Max_missing_count")),
  trainDF.columns.find(_ == "origin_weather_missing_h7"),
  trainDF.columns.find(_ == "destination_weather_missing_h7")
).flatten

println("\n📊 Échantillon de données (5 premières lignes):")
println("Colonnes sélectionnées:")
sampleCols.foreach(c => println(s"  - $c"))
println()

trainDF.select(sampleCols.map(col): _*).show(5, truncate = false)
```

## ✅ 7. Vérification Finale : Aucun NULL

```scala
// Vérifier qu'il n'y a plus de NULL (TRAIN)
val columnsWithNullsTrain = trainDF.columns.flatMap { colName =>
  val nullCount = trainDF.filter(col(colName).isNull).count()
  if (nullCount > 0) {
    val nullPct = (nullCount.toDouble / trainCount * 100)
    Some((colName, nullCount, nullPct))
  } else {
    None
  }
}.sortBy(-_._2)

println("\n🔍 Vérification de l'absence de NULL (TRAIN):")
println("=" * 80)

if (columnsWithNullsTrain.isEmpty) {
  println("✅ SUCCÈS: Aucune valeur NULL détectée dans aucune colonne !")
  println("✅ Tous les NULL ont été remplacés par des sentinelles")
} else {
  println("⚠️  ATTENTION: Des valeurs NULL ont été détectées:")
  columnsWithNullsTrain.foreach { case (colName, count, pct) =>
    println(f"$colName%-60s: $count%,8d ($pct%5.2f%%)")
  }
  println("\n⚠️  Il reste des NULL - vérifier le pipeline de traitement")
}
```

```scala
// Vérifier qu'il n'y a plus de NULL (TEST)
val columnsWithNullsTest = testDF.columns.flatMap { colName =>
  val nullCount = testDF.filter(col(colName).isNull).count()
  if (nullCount > 0) {
    val nullPct = (nullCount.toDouble / testCount * 100)
    Some((colName, nullCount, nullPct))
  } else {
    None
  }
}.sortBy(-_._2)

println("\n🔍 Vérification de l'absence de NULL (TEST):")
println("=" * 80)

if (columnsWithNullsTest.isEmpty) {
  println("✅ SUCCÈS: Aucune valeur NULL détectée dans aucune colonne !")
  println("✅ Tous les NULL ont été remplacés par des sentinelles")
} else {
  println("⚠️  ATTENTION: Des valeurs NULL ont été détectées:")
  columnsWithNullsTest.foreach { case (colName, count, pct) =>
    println(f"$colName%-60s: $count%,8d ($pct%5.2f%%)")
  }
  println("\n⚠️  Il reste des NULL - vérifier le pipeline de traitement")
}
```

## 📊 8. Résumé Final

```scala
println("\n" + "=" * 80)
println("                      RÉSUMÉ DE LA VALIDATION")
println("=" * 80)

println(f"\n✓ Datasets chargés:")
println(f"  - TRAIN: $trainCount%,10d lignes, ${trainDF.columns.length}%4d colonnes")
println(f"  - TEST:  $testCount%,10d lignes, ${testDF.columns.length}%4d colonnes")

println(f"\n✓ Flags par heure:           ${hourlyFlagCols.length}%4d colonnes")
println(f"✓ Features agrégées:         ${aggFeatureCols.length}%4d colonnes")
println(f"✓ Compteurs missing:         ${counterCols.length}%4d colonnes")
println(f"\n✓ Sentinelles numériques:    ${colsWithSentinels.length}%4d colonnes avec -999")
println(f"✓ Sentinelles string:        ${colsWithMissing.length}%4d colonnes avec MISSING")

val trainNullStatus = if (columnsWithNullsTrain.isEmpty) "✅ Aucun NULL" else s"⚠️  ${columnsWithNullsTrain.length} colonnes avec NULL"
val testNullStatus = if (columnsWithNullsTest.isEmpty) "✅ Aucun NULL" else s"⚠️  ${columnsWithNullsTest.length} colonnes avec NULL"

println(f"\n✓ Validation NULL TRAIN:     $trainNullStatus")
println(f"✓ Validation NULL TEST:      $testNullStatus")

println("\n" + "=" * 80)

if (columnsWithNullsTrain.isEmpty && columnsWithNullsTest.isEmpty) {
  println("\n🎉 VALIDATION RÉUSSIE: Tous les NULL ont été correctement traités !")
} else {
  println("\n⚠️  VALIDATION PARTIELLE: Certains NULL subsistent - vérifier le pipeline")
}
```
